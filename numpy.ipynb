```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin)
plt.plot(x, y_cos)
plt.xlabel('lda')
plt.ylabel('Vo Iu')
plt.title('anh iu em')
plt.legend(['anh','em'])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.subplot(2, 1 ,1) #plt.subplot(nrows, ncols, index), so hang, so cot, index cua bieu do
plt.plot(x, y_sin)
plt.title('sine')

plt.subplot(2,1,2)
plt.plot(x, y_cos)
plt.title('cosine')
plt.show()
```
